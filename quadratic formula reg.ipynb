```python
import numpy as np
import pandas as pd
from numpy import linalg
import math
import matplotlib.pyplot as plt
```

```python
a = 4
b = 5
c = 4

X = np.linspace(1, 10, 150)

y_0 = a * X**2 + b * X + c
```

```python
plt.plot(X, y_0)
```

```python
uglify = np.random.uniform(-25,25,size=150)
uglify
```

```python
y = y_0 + uglify
plt.plot(X,y_0)
plt.scatter(X, y)
plt.show()
```

```python
A = np.array([[sum(X**4),sum(X**3),sum(X**2)],
              [sum(X**3),sum(X**2),sum(X)],
              [sum(X**2),sum(X**2),len(X)]])
b = [sum((X**2)*y),sum(X*y),sum(y)]
```

```python
coeff = linalg.solve(A,b)
y_n = coeff[0]*X**2 + coeff[1]*X + coeff[2]
plt.plot(X,y_n,color = 'r', label = "y_new = {:.2f}x² + {:.2f}x + {:.2f}".format(coeff[0],coeff[1],coeff[2]))
plt.scatter(X, y)
plt.legend()
plt.plot(X,y_0,color = 'g')
print(coeff)
```

```python
#SSE = sum(y-y_pred)**2
#SST = sum(y-y_mean)**2

SSE = sum((y-y_n)**2)
SST = sum((y-np.mean(y))**2)

r_squared = 1 - (SSE/SST)
print(r_squared)
```

# Numpy

```python
a_np, b_np, c_np = np.polyfit(X, y, 2)

y_np = a_np*X**2 + b_np*X + c_np

print(f"y_np = {a_np:.2f}x² + {b_np:.2f}x + {c_np:.2f}")
```

```python
r_squared_np = 1 - (sum((y-y_np)**2)/sum((y-np.mean(y))**2))
print("r_squared = ", r_squared)
print("r_squared_np = ", r_squared_np)
```
