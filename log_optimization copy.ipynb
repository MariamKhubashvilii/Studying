```python
import numpy as np
import pandas as pd
from numpy import linalg
import matplotlib.pyplot as plt
```

# აღწერა

y = A*e^Bx

```python
A = 1.5
B = 1.6
n = 200

X = np.linspace(1, 2, n)

y_0 = A * np.e**(B * X)
```

```python
uglify = np.random.uniform(-1,1,size = n)

plt.plot(X, y_0)
plt.show()
```

```python
y = y_0 + uglify
plt.plot(X,y_0)
plt.scatter(X, y)
plt.show()
```

# წრფივ დამოკიდებულებად გარდაქმნა

lny = ln(A*e^Bx), lny = lnA + ln(e^Bx), lny = lnA + Bx, My coefficients: lnA; B

```python
y = np.log(y)

#y = np.log(A) + B * X

plt.scatter(X, y, s=1)
plt.show()
```

```python
A = np.array([[sum(X*X),sum(X)],
              [sum(X),n]])
b = [sum(X*y),sum(y)]
```

```python
coeff = linalg.solve(A,b)
y_n = coeff[0]*X + coeff[1]
plt.plot(X,y_n,color = 'r', label = "y_new = {:.2f}x + {:.2f}".format(coeff[0],coeff[1]))
plt.scatter(X, y_n)
plt.legend()
plt.plot(X,y,color = 'g')
print(coeff)
```

```python
A = np.exp(coeff[1])
B = coeff[0]

y = A * np.e**(B * X)
plt.plot(X,y_0,color = 'r', label = "y_new = {:.2f}e^({:.2f}x)".format(A,B))
plt.scatter(X, y)
plt.legend()
```
