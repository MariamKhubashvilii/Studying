თეთრ ფონზე შავი ჯვარი (მართკუთხედში)

```python
black_cross = np.array([[0]*m for i in range(n)])
```

```python
half_n = n // 2 + 1 
half_m = m // 2 + 1
```

```python
for i in range(n):
    for j in range(m):
        if i == half_n or j == half_m:
            black_cross[i - 1][j - 1] = 1
```

```python
plt.imshow(black_cross, cmap='binary', vmin=0, vmax=1)
plt.axis('off')
plt.show()
```

შავ ფონზე თეთრი ჯვარი (მართკუთხედში)

```python
white_cross = np.array([[1]*m for i in range(n)])

half_n = n // 2 + 1 
half_m = m // 2 + 1

for i in range(n):
    for j in range(m):
        if i == half_n or j == half_m:
            white_cross[i - 1][j - 1] = 0

plt.imshow(white_cross, cmap='binary', vmin=0, vmax=1)
plt.axis('off')
plt.show()
```

შავ ფონზე დიაგონალური ჯვარი (კვადრატში)

```python
diagonal_cross_sq = np.array([[1]*n for i in range(n)])
```

```python
for i in range(n):
    for j in range(n):
        if i == j:
            diagonal_cross_sq[i][j] = 0
        if i+j == n - 1:
            diagonal_cross_sq[i][j] = 0
```

```python
plt.imshow(diagonal_cross_sq, cmap='binary', vmin=0, vmax=1)
plt.axis('off')
plt.show()
```

შავ ფონზე თეთრი დიაგონალური ჯვარი (მართკუთხედში)

```python
diagonal_cross = np.array([[1]*m for i in range(n)])
```

```python
x = np.linspace(0,m,1000)

y = (-n/m)*x + n

y1 = (n/m)*x
diagonal_points = np.array([x,y])

plt.scatter(x,y)
```

```python
diagonal_points
```

```python
main_points = np.array([x, y])
anti_points = np.array([x, y1])

for points in [main_points, anti_points]:
    for j in range(1, m + 1):
        for i in range(1, n + 1):
            mask = ((points[0] >= j - 1) & (points[0] < j) &
                    (points[1] >= i - 1) & (points[1] < i))
            if np.sum(mask) > 0:
                diagonal_cross[i - 1][j - 1] = 0
```

```python
width_px = 1000
height_px = 1500
dpi = 100
fig = plt.figure(figsize=(width_px / dpi, height_px / dpi), dpi=dpi)

plt.imshow(diagonal_cross, cmap='binary', vmin=0, vmax=1)

plt.axis('off')
plt.show()
```
