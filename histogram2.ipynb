```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 10000
bins = 100
x1 = np.random.uniform(5,25,n)
y1 = np.random.uniform(4,16,n)
```

```python
plt.scatter(x1, y1, s = 1)
```

```python
uniform_hist_x1 = np.histogram(x1, bins)
plt.scatter(np.linspace(0,100,100), uniform_hist_x1[0])
plt.ylim(0,max(uniform_hist_x1[0]) + 10)
```

```python
uniform_hist_y1 = np.histogram(y1, bins)
plt.scatter(np.linspace(0,100,100), uniform_hist_y1[0])
plt.ylim(0, 140)
```
