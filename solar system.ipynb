```python
import numpy as np
import matplotlib.pyplot as plt
from ipyevents import Event
from ipywidgets import Output
from IPython.display import display
```

```python
#circle
circle_r = 3
#elipse 1
a1 = 8
b1 = 5
#elipse 2
a2 = 6
b2 = 5

#celestial bodies
solar_radius = 0.3
constant = 2.5
time_interval = 5

#angle
theta = np.linspace(-10,10,100)
alpha = np.pi*45/180
OF = np.sqrt(a1**2 - b1**2)

#specific angle
null_angle = 0

#coordinates of the circle
x_circle = circle_r*np.cos(theta)
y_circle = circle_r*np.sin(theta)
revolution_circle = np.sqrt(circle_r**3*constant)
planet_circle = null_angle + time_interval/revolution_circle

#planet
#x_pl_circle = circle_r*np.cos()
#y_pl_circle = circle_r*np.random(np.sin(theta))

#rotate the flat ellipse by alpha, then shift the focus (distance OF) onto the sun
x_el_1 = a1*np.cos(theta)*np.cos(alpha) - b1*np.sin(theta)*np.sin(alpha) - OF*np.cos(alpha)
y_el_1 = a1*np.cos(theta)*np.sin(alpha) + b1*np.sin(theta)*np.cos(alpha) - OF*np.sin(alpha)
revolution_el1 = a1**3*constant

#coordinates of the second ellipse

y_el_2 = b2*np.cos(theta) 
x_el_2 = a2*np.sin(theta) + (a2**2-b2**2)**(1/2)
revolution_el2 = a2**3*constant
```

```python
fig, ax = plt.subplots()

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)

plt.plot(x_circle, y_circle)
plt.plot(x_el_1, y_el_1)
plt.plot(x_el_2, y_el_2)

sun = plt.Circle((0,0), solar_radius, color = "y")

ax.add_patch(sun)

plt.axis('equal')
```
