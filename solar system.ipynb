```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#circle
circle_r = 3
#elipse 1
a1 = 8
b1 = 5
#elipse 2
a2 = 6
b2 = 5

#celestial bodies
solar_radius = 0.3

#angle
theta = np.linspace(-10,10,100)
alpha = 2

#specific angle
theta_rand = np.random.choice(theta)

#coordinates of the circle
x_circle = circle_r*np.cos(theta)
y_circle = circle_r*np.sin(theta)

#planet
#x_pl_circle = circle_r*np.cos()
#y_pl_circle = circle_r*np.random(np.sin(theta))

#coordinates of the first ellipse
rad_vec_1 = (a1**2 - b1**2)**(1/2)
x_el_1 = a1*np.cos(theta+alpha) + rad_vec_1*np.sin(alpha)
y_el_1 = b1*np.sin(theta) + rad_vec_1*np.cos(alpha)


#coordinates of the second ellipse

y_el_2 = b2*np.cos(theta)
x_el_2 = a2*np.sin(theta) + a2 - (a2**2-b2**2)**(1/2)

```

```python
fig, ax = plt.subplots()

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)

plt.plot(x_circle, y_circle)
plt.plot(x_el_1, y_el_1)
plt.plot(x_el_2, y_el_2)

sun = plt.Circle((0,0), solar_radius, color = "y")

ax.add_patch(sun)

plt.axis('equal')
```
