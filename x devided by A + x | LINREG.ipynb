```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = 5
B = 2
n = 200
x = np.linspace(2, 50, n)
y_0 = A * x / (B + x)

plt.plot(x, y_0)
plt.axhline(A, color="gray", linestyle="--")
plt.xlim(0, 50)
plt.ylim(0, 6)
plt.show()
```

```python
uglify = np.random.uniform(-0.5, 0.5, size=n)
y = y_0 + uglify
```

```python
plt.scatter(x, y)
plt.axhline(A, color="gray", linestyle="--")
plt.xlim(0, 50)
plt.ylim(0, 6)
plt.show()
```

```python
X = 1/x
Y = 1/y
```

```python
mat = np.array([[sum(X*X),sum(X)],
              [sum(X),len(X)]])
matb = [sum(X*Y),sum(Y)]
```

```python
coeff = np.linalg.solve(mat,matb)
A = 1/coeff[1]
B = coeff[0]/coeff[1]
```

```python
y_pred = A * x / (B + x)

SSE = sum((y-y_pred)**2)
SST = sum((y-np.mean(y))**2)

r_squared = 1 - (SSE/SST)
print(r_squared)

plt.plot(x,y_pred,color = 'r', label = "y_new = {:.2f}x + {:.2f}, R² = {:.2f}".format(A, B, r_squared))
plt.scatter(x, y)
plt.legend()
plt.plot(x,y_0,color = 'g')
print(coeff)
```

```python
slope, intercept = np.polyfit(1/x, 1/y, 1)

y_np = slope*X + intercept

plot = plt.plot(x, 1/y_np, color='orange', label="y_new = {:.2f}x + {:.2f}, R² = {:.2f}".format(A, B, r_squared))
```
