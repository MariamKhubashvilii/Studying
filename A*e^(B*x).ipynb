```python
import numpy as np
import pandas as pd
from numpy import linalg
import matplotlib.pyplot as plt
```

# აღწერა

y = A*x^B

```python
A = 1.5
B = 1.6
n = 200

X = np.linspace(4, 6, n)

y_0 = A * X**B
```

```python
uglify = np.random.uniform(-4,4,size = n)

plt.plot(X, y_0)
plt.show()
```

```python
y = y_0 + uglify
plt.plot(X,y_0)
plt.scatter(X, y)
plt.show()
```

# წრფივ დამოკიდებულებად გარდაქმნა

```python
y_log = np.log10(y)
x_log = np.log10(X)
#y = np.log(A) + B * X

plt.scatter(x_log, y_log, s=1)
plt.show()
```

```python
A = np.array([[sum(x_log*x_log),sum(x_log)],
              [sum(x_log),n]])
b = [sum(x_log*y_log),sum(y_log)]
```

```python
coeff = linalg.solve(A,b)
y_n = coeff[0]*x_log + coeff[1]
plt.plot(x_log,y_n,color = 'r', label = "y_new = {:.2f}x + {:.2f}".format(coeff[0],coeff[1]))
plt.legend()
plt.scatter(x_log,y_log,color = 'g')
print(coeff)
```

```python
A = 10**(coeff[1])
B = coeff[0]

y = A * X**B
plt.plot(X,y,color = 'r', label = "y_new = {:.2f}x^({:.2f})".format(A,B))
plt.scatter(X, y)
plt.legend()
```

```python
from scipy.optimize import curve_fit
```

```python
parameters, covariance = curve_fit(lambda x, a, b: a * x**b, X, y, p0=[1, 1])
```
