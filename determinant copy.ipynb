```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from numpy import linalg
```

```python
matrix = np.array([[ 2, 3, -1],
              [ 4, 1,  2],
              [-2, 5,  3]], dtype=float)
bA = np.array([5, 6, 7], dtype=float)
```

```python
def det3x3(matrix):
    '''Sarrus'''
    s1 = s2 = s3 = 1
    m1 = m2 = m3 = 1
    for i in range(3):
        s1 *= matrix[i,i]
        s2 *= matrix[i, (i + 1) % 3]
        s3 *= matrix[i, (i + 2) % 3]
    for i in range(3):
        m1 *= matrix[i, - (i + 1) % 3]
        m2 *= matrix[i, (1 - i) % 3]
        m3 *= matrix[i, -i % 3]
    d = s1 + s2 + s3 - m1 - m2 - m3
    return d
```

```python
determinant = det3x3(matrix)
determinants = []
for i in range(3):
    '''Cramer'''
    temp = matrix.copy()
    temp[:, i] = bA
    determinants.append(det3x3(temp))

print("Determinant of A:", determinant, 
      "\nDeterminants of modified matrices:", determinants)
```

```python
x = 0
for i in range(3):
    global x
    x = determinants[i] / determinant
    print(f"x_{i+1} = {x}")
```

# კონკრეტული

```python
# 2x + 3y - z = 5
# 4x +  y + 2z = 6
# -2x + 5y + 3z = 7

A = np.array([[ 2, 3, -1],
              [ 4, 1,  2],
              [-2, 5,  3]], dtype=float)
b = np.array([5, 6, 7], dtype=float)

x = linalg.solve(A, b)

print("ამონახსნები:", x)
print(np.allclose(A @ x, b))
```
