```python
# Draw center X and Y axes (at 0, 0)
plt.axhline(0, color="black", linewidth=1, linestyle="-")  # X-axis line
plt.axvline(0, color="black", linewidth=1, linestyle="-")  # Y-axis line

# Mark the center point (0, 0)
plt.plot(0, 0, marker="o", markersize=8, color="red", label="Center (0,0)")

# Add grid and legend
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
